O desafio proposto é: gerar um circuito quantico de 2 qubits que tenha como saída os 4 possíveis estados com fases separadas de pi/4.

- 00 -> fase 0
- 01 -> fase de pi/4
- 10 -> fase de pi/2
- 11 -> fase de 3pi/4


```python
from math import pi, gcd #(util para o algoritmo de Shor)

#apemas para operacoes e representacoes matemáticas
from fractions import Fraction
import numpy as np

#qiskit
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister, transpile
from qiskit.visualization import plot_histogram
from qiskit.quantum_info import Statevector
from qiskit.visualization import array_to_latex

try:
    from qiskit_aer import AerSimulator
    backend = AerSimulator()
    print(":-)")
except Exception:
    backend = None
    print(":-(")

print("Importações carregadas!")
```

```python
# Inicializando o circuito quântico
n = 2
qc = QuantumCircuit(n, n)
```

```python
qc.h(0)
qc.h(1)

qc.barrier()
```

```python
qc.draw('mpl')
```

```python
qc.p(pi/4, 0)
qc.p(pi/2, 1)

qc.barrier()
```

```python
qc.draw('mpl')
```

```python
# Estado final
sv = Statevector.from_instruction(qc)

# Transforma em vetor-coluna, por estética
sv_coluna = np.asarray(sv).reshape(-1, 1)

display(array_to_latex(sv_coluna, precision=2, max_size=100))
```

## Separando módulos e fases
Os módulos devem ser todos 0.5 (probabilidade de 25% cada) e as fases 0, π/4, π/2 e 3π/4 (≈ 0, 0.79, 1.57 e 2.36 rad).

```python
display(array_to_latex(np.abs(sv_coluna), precision=2, max_size=100))  # módulo dos estados
display(array_to_latex(np.angle(sv_coluna), precision=2, max_size=100))  # fase dos estados
```
