O desafio proposto agora é o mesmo para 3 qubits: gerar um circuito quantico de 3 qubits que tenha como saída os 8 possíveis estados com fases separadas de pi/4

- 000 -> fase 0
- 001 -> fase de pi/4
- 010 -> fase de pi/2
- 011 -> fase de 3pi/4
- 100 -> fase pi
- 101 -> fase de 5pi/4
- 110 -> fase de 6i/2
- 111 -> fase de 7pi/4


```python
from math import pi, gcd #(util para o algoritmo de Shor)

#apemas para operacoes e representacoes matemáticas
from fractions import Fraction
import numpy as np

#qiskit
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister, transpile
from qiskit.visualization import plot_histogram
from qiskit.quantum_info import Statevector
from qiskit.visualization import array_to_latex

try:
    from qiskit_aer import AerSimulator
    backend = AerSimulator()
    print(":-)")
except Exception:
    backend = None
    print(":-(")

print("Importações carregadas!")
```

```python
# inicializando o circuito
n = 3
qc = QuantumCircuit(n, n)
```

```python
qc.h(range(n))

qc.barrier()
```

```python
qc.draw('mpl')
```

```python
qc.p(pi/4, 0)
qc.p(pi/2, 1)
qc.p(pi, 2)

qc.barrier()
```

```python
qc.draw('mpl')
```

```python
# Estado final
sv = Statevector.from_instruction(qc)

# Transforma em vetor-coluna, por estética
sv_coluna = np.asarray(sv).reshape(-1, 1)

display(array_to_latex(sv_coluna, precision=2, max_size=100))
```

```python
display(array_to_latex(np.abs(sv_coluna), precision=2, max_size=100))  # módulo dos estados
display(array_to_latex(np.angle(sv_coluna), precision=2, max_size=100))  # fase dos estados
```
